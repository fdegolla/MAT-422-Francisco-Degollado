## 2.3.1 Joint Probability Distributions

A joint probability distribution describes the probability that two random variables take specific values at the same time.

For two discrete random variables $X$ and $Y$, the joint probability is

$$
P(X=x,Y=y).
$$

Consider flipping two fair coins. Let $X$ represent the result of the first coin and $Y$ represent the result of the second coin, where

- $0$ represents tails.
- $1$ represents heads.

The sample space is

$$
S=\{(0,0),(0,1),(1,0),(1,1)\}.
$$

Since the coins are fair and independent, each outcome has a probability of $0.25$.

We can use Python to calculate the joint probability distribution and verify that the probabilities sum to $1$.


```python

# Define the possible outcomes
X = [0, 1]
Y = [0, 1]

# Create the joint probability distribution
joint_prob = np.array([
    [0.25, 0.25],
    [0.25, 0.25]
])

print("Joint probability distribution:")
print(joint_prob)

# Calculate the total probability
total_probability = np.sum(joint_prob)

print("\nTotal probability:", total_probability)

# Verify that the probabilities sum to 1
print("Verification:", np.isclose(total_probability, 1))
```

### Interpretation

The joint probability distribution gives the probability of each possible combination of $X$ and $Y$. Since the two coins are fair and independent, each of the four possible outcomes has probability $0.25$. The probabilities sum to $1$, and the verification returns `True`, confirming that this is a valid joint probability distribution.

## 2.3.2 Correlation and Dependence

Correlation measures the strength and direction of the linear relationship between two random variables.

The correlation coefficient is

$$
\rho_{X,Y}=\frac{\operatorname{Cov}(X,Y)}
{\sigma_X\sigma_Y}.
$$

The correlation coefficient ranges from $-1$ to $1$:

- A value close to $1$ indicates a strong positive linear relationship.
- A value close to $-1$ indicates a strong negative linear relationship.
- A value close to $0$ indicates little or no linear relationship.

Consider the following paired data:

$$
X=\{1,2,3,4,5\}
$$

and

$$
Y=\{2,4,5,8,10\}.
$$

We can use Python to calculate the correlation coefficient and examine the relationship between $X$ and $Y$.

```python
# Define the two variables
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 5, 8, 10])

# Calculate the correlation matrix
correlation_matrix = np.corrcoef(X, Y)

# Extract the correlation coefficient
correlation = correlation_matrix[0, 1]

print("X =", X)
print("Y =", Y)

print("\nCorrelation matrix:")
print(np.round(correlation_matrix, 4))

print("\nCorrelation coefficient =", round(correlation, 4))
```

```python
# Create a scatter plot
plt.scatter(X, Y)

# Add labels and title
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Correlation Between X and Y")

# Add grid
plt.grid()

plt.show()
```

### Interpretation

The correlation coefficient between $X$ and $Y$ is approximately $0.9901$. Since this value is close to $1$, the variables have a strong positive linear relationship. The scatter plot also shows that as $X$ increases, $Y$ generally increases. Therefore, these two variables are strongly correlated and are not independent in this example.

## 2.3.3 Random Samples

A random sample is a collection of observations selected randomly from a population. Random sampling allows us to use information from a sample to estimate characteristics of the population.

Suppose a population follows a normal distribution with mean

$$
\mu = 50
$$

and standard deviation

$$
\sigma = 10.
$$

We can generate a random sample of $100$ observations from this population and calculate the sample mean and sample standard deviation.

The sample mean is

$$
\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i.
$$

We expect the sample statistics to be reasonably close to the population parameters, although they will not usually be exactly equal.

```python
# Set the random seed so the results are reproducible
np.random.seed(42)

# Define the population parameters
mu = 50
sigma = 10

# Generate a random sample of 100 observations
sample = np.random.normal(mu, sigma, 100)

# Calculate sample statistics
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

print("Sample size =", len(sample))
print("Population mean =", mu)
print("Sample mean =", round(sample_mean, 4))

print("\nPopulation standard deviation =", sigma)
print("Sample standard deviation =", round(sample_std, 4))
```

```python
# Create a histogram of the random sample
plt.hist(sample, bins=10, edgecolor="black")

# Add labels and title
plt.xlabel("Sample Value")
plt.ylabel("Frequency")
plt.title("Histogram of Random Sample")

# Show the population mean
plt.axvline(mu, linestyle="--", label="Population Mean")

plt.legend()
plt.grid()

plt.show()
```

### Interpretation

A random sample of $100$ observations was generated from a normal population with mean $\mu=50$ and standard deviation $\sigma=10$. The sample mean is approximately $48.9615$ and the sample standard deviation is approximately $9.0817$. These values are close to the population parameters but are not exactly equal because of random sampling variability. The histogram shows that the observations are distributed around the population mean of $50$. This demonstrates how a random sample can be used to estimate characteristics of a population.
