## 2.4.1 MLE for Random Samples

Maximum Likelihood Estimation (MLE) is a method for estimating unknown parameters of a probability distribution using observed sample data.

Suppose we have a random sample

$$
X_1,X_2,\ldots,X_n
$$

from a normal distribution with unknown mean $\mu$ and variance $\sigma^2$.

For a normal distribution, the maximum likelihood estimate of the mean is

$$
\hat{\mu}=\frac{1}{n}\sum_{i=1}^{n}X_i,
$$

and the maximum likelihood estimate of the variance is

$$
\hat{\sigma}^2=\frac{1}{n}\sum_{i=1}^{n}(X_i-\hat{\mu})^2.
$$

Consider the following random sample:

$$
X=\{12,15,14,10,13,16,14,15,11,13\}.
$$

We can use Python to calculate the maximum likelihood estimates of the mean and variance.

```python
# Define the random sample
sample = np.array([12, 15, 14, 10, 13, 16, 14, 15, 11, 13])

# Number of observations
n = len(sample)

# MLE of the mean
mu_hat = np.mean(sample)

# MLE of the variance
variance_hat = np.mean((sample - mu_hat)**2)

# MLE of the standard deviation
sigma_hat = np.sqrt(variance_hat)

print("Sample =", sample)
print("Sample size =", n)

print("\nMLE of mean =", mu_hat)
print("MLE of variance =", variance_hat)
print("MLE of standard deviation =", sigma_hat)
```

### Interpretation

Using maximum likelihood estimation, the estimated population mean is $\hat{\mu}=13.3$ and the estimated population variance is $\hat{\sigma}^2=3.21$. The estimated standard deviation is approximately $1.7916$. These values are the parameter estimates that make the observed random sample most likely under the assumed normal distribution.

## 2.4.2 Linear Regression

Linear regression models the relationship between a dependent variable $Y$ and an independent variable $X$ using a straight line.

The simple linear regression model is

$$
Y = \beta_0 + \beta_1 X + \epsilon,
$$

where $\beta_0$ is the y-intercept, $\beta_1$ is the slope, and $\epsilon$ represents random error.

The estimated regression line is

$$
\hat{Y} = \hat{\beta}_0 + \hat{\beta}_1 X.
$$

Consider the following data:

$$
X=\{1,2,3,4,5,6\}
$$

and

$$
Y=\{2,4,5,4,7,8\}.
$$

We can use Python to estimate the slope and intercept of the regression line and use the resulting model to predict values of $Y$.

```python
# Define the data
X = np.array([1, 2, 3, 4, 5, 6])
Y = np.array([2, 4, 5, 4, 7, 8])

# Calculate the linear regression coefficients
slope, intercept = np.polyfit(X, Y, 1)

# Calculate predicted Y values
Y_predicted = slope * X + intercept

print("Slope =", round(slope, 4))
print("Y-intercept =", round(intercept, 4))

print("\nRegression equation:")
print("Y_hat =", round(intercept, 4), "+", round(slope, 4), "X")

print("\nPredicted Y values:")
print(np.round(Y_predicted, 4))
```

```python
# Plot the original data points
plt.scatter(X, Y, label="Data points")

# Plot the regression line
plt.plot(X, Y_predicted, label="Regression line")

# Add labels and title
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")

# Add legend and grid
plt.legend()
plt.grid()

plt.show()
```

### Interpretation

The fitted linear regression equation is approximately

$$
\hat{Y} = 1.2 + 1.0857X.
$$

The positive slope indicates that $Y$ generally increases as $X$ increases. Specifically, the model predicts that $Y$ increases by approximately $1.0857$ units for every one-unit increase in $X$. The graph shows that the regression line follows the overall trend of the data, although the individual data points do not all lie exactly on the line.
