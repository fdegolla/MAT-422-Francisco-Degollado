## 2.2.1 Probability Axioms

Probability measures the likelihood that an event will occur. The probability axioms state that:

1. $P(A) \geq 0$ for any event $A$.
2. $P(S) = 1$, where $S$ is the entire sample space.
3. For mutually exclusive events $A$ and $B$,

$$
P(A \cup B) = P(A) + P(B).
$$

Consider rolling a fair six-sided die. The sample space is

$$
S = \{1,2,3,4,5,6\}.
$$

Let $A = \{2,4,6\}$ be the event of rolling an even number and
$B = \{1,3,5\}$ be the event of rolling an odd number.

Since $A$ and $B$ are mutually exclusive, we can use Python to calculate their probabilities and verify the probability axioms.

```python
# Define the sample space
S = {1, 2, 3, 4, 5, 6}

# Define the events
A = {2, 4, 6}  # Even numbers
B = {1, 3, 5}  # Odd numbers

# Calculate probabilities
P_A = len(A) / len(S)
P_B = len(B) / len(S)

# Find the union and intersection
A_union_B = A.union(B)
A_intersection_B = A.intersection(B)

P_union = len(A_union_B) / len(S)

print("P(A) =", P_A)
print("P(B) =", P_B)

print("\nA intersection B =", A_intersection_B)
print("A union B =", A_union_B)

print("\nP(A union B) =", P_union)
print("P(A) + P(B) =", P_A + P_B)

# Verify the probability axiom
print("\nAddition axiom verified:", np.isclose(P_union, P_A + P_B))
```

### Interpretation

Events $A$ and $B$ each have probability $0.5$. Their intersection is empty, so the events are mutually exclusive. Their union contains the entire sample space, giving $P(A \cup B)=1$. Since $P(A)+P(B)=0.5+0.5=1$, the addition axiom is verified.

## 2.2.2 Conditional Probability

Conditional probability measures the probability that an event $A$ occurs given that another event $B$ has already occurred. It is defined as

$$
P(A \mid B)=\frac{P(A \cap B)}{P(B)},
$$

where $P(B)>0$.

Consider rolling a fair six-sided die with sample space

$$
S=\{1,2,3,4,5,6\}.
$$

Let $A=\{2,4,6\}$ be the event of rolling an even number and let $B=\{4,5,6\}$ be the event of rolling a number greater than 3.

The intersection is

$$
A \cap B=\{4,6\}.
$$

We can use Python to calculate $P(A \mid B)$ and verify the conditional probability formula.

```python
# Define the sample space
S = {1, 2, 3, 4, 5, 6}

# Define the events
A = {2, 4, 6}  # Even numbers
B = {4, 5, 6}  # Numbers greater than 3

# Find the intersection
A_intersection_B = A.intersection(B)

# Calculate probabilities
P_B = len(B) / len(S)
P_intersection = len(A_intersection_B) / len(S)

# Calculate conditional probability
P_A_given_B = P_intersection / P_B

print("Event A =", A)
print("Event B =", B)
print("A intersection B =", A_intersection_B)

print("\nP(B) =", P_B)
print("P(A intersection B) =", P_intersection)

print("\nP(A | B) =", P_A_given_B)
```

### Interpretation

The probability of rolling a number greater than 3 is $P(B)=0.5$. The probability of rolling a number that is both even and greater than 3 is $P(A \cap B)=1/3$. Using the conditional probability formula gives $P(A \mid B)=2/3 \approx 0.667$. Therefore, given that the die roll is greater than 3, there is approximately a $66.7\%$ chance that the number is even.

## 2.2.3 Discrete Random Variables

A discrete random variable takes values from a finite or countable set. Each possible value has an associated probability described by a probability mass function (PMF).

Consider a fair six-sided die. Let $X$ represent the number rolled. The possible values are

$$
X=\{1,2,3,4,5,6\}.
$$

Since the die is fair, each outcome has probability

$$
P(X=x)=\frac{1}{6}.
$$

The expected value of a discrete random variable is

$$
E[X]=\sum_x xP(X=x),
$$

and the variance is

$$
Var(X)=\sum_x (x-E[X])^2P(X=x).
$$

We can use Python to calculate the probability distribution, expected value, and variance of $X$.

```python
# Possible values of the discrete random variable
x = np.array([1, 2, 3, 4, 5, 6])

# Probability of each outcome
probabilities = np.array([1/6] * 6)

# Calculate the expected value
expected_value = np.sum(x * probabilities)

# Calculate the variance
variance = np.sum((x - expected_value)**2 * probabilities)

print("Possible values:", x)
print("Probabilities:", np.round(probabilities, 4))

print("\nSum of probabilities:", np.sum(probabilities))
print("Expected value E[X] =", expected_value)
print("Variance Var(X) =", variance)
```

### Interpretation

The probabilities of all six possible outcomes sum to approximately $1$, as required for a probability distribution. The expected value is $E[X]=3.5$, which represents the long-run average value of many die rolls. The variance is approximately $2.917$, which measures how spread out the possible outcomes are around the expected value.

## 2.2.4 Continuous Random Variables

A continuous random variable can take any value within an interval. Unlike a discrete random variable, probabilities are described by a probability density function (PDF).

For a normal random variable with mean $\mu$ and standard deviation $\sigma$, the probability density function is

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}.
$$

For a standard normal distribution,

$$
\mu=0, \qquad \sigma=1.
$$

The probability that $X$ falls between two values $a$ and $b$ is the area under the probability density function:

$$
P(a \leq X \leq b)=\int_a^b f(x)\,dx.
$$

We can use Python to graph the standard normal distribution and calculate the probability that $X$ falls between $-1$ and $1$.

```python
from scipy.stats import norm

# Define the standard normal distribution
mu = 0
sigma = 1

# Calculate P(-1 <= X <= 1)
probability = norm.cdf(1, mu, sigma) - norm.cdf(-1, mu, sigma)

print("Mean =", mu)
print("Standard deviation =", sigma)
print("P(-1 <= X <= 1) =", probability)

# Create values for the normal distribution
x_values = np.linspace(-4, 4, 500)
pdf_values = norm.pdf(x_values, mu, sigma)

# Plot the probability density function
plt.plot(x_values, pdf_values)

# Shade the area between -1 and 1
x_fill = np.linspace(-1, 1, 200)
plt.fill_between(
    x_fill,
    norm.pdf(x_fill, mu, sigma),
    alpha=0.3
)

plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("Standard Normal Distribution")
plt.grid()

plt.show()
```

### Interpretation

The standard normal distribution has a mean of $\mu=0$ and a standard deviation of $\sigma=1$. The probability that $X$ falls between $-1$ and $1$ is approximately $0.6827$, or $68.27\%$. The shaded region of the graph represents this probability as the area under the probability density function between $-1$ and $1$. This demonstrates how probabilities for continuous random variables are found using areas under a probability density function.
