## 1.2.1 Linear Spaces

A linear space, also called a vector space, is a set of vectors that is closed under vector addition and scalar multiplication.

Consider the following two vectors in $\mathbb{R}^2$:

$$
v_1 =
\begin{bmatrix}
1\\
2
\end{bmatrix},
\qquad
v_2 =
\begin{bmatrix}
3\\
1
\end{bmatrix}
$$

We can demonstrate closure by adding the two vectors and by multiplying a vector by a scalar.

```python
import numpy as np

# Define two vectors in R^2
v1 = np.array([1, 2])
v2 = np.array([3, 1])

# Add the two vectors
vector_sum = v1 + v2

# Multiply v1 by the scalar 2
scalar_multiple = 2 * v1

print("v1 + v2 =", vector_sum)
print("2v1 =", scalar_multiple)
```

### Interpretation

The sum $v_1 + v_2 = (4,3)$ is still a vector in $\mathbb{R}^2$. The scalar multiple $2v_1 = (2,4)$ is also in $\mathbb{R}^2$. Therefore, this example demonstrates closure under vector addition and scalar multiplication.

## 1.2.2 Orthogonality

Two vectors are orthogonal if their dot product is equal to zero.

Consider the following two vectors in $\mathbb{R}^2$:

$$
u =
\begin{bmatrix}
1\\
2
\end{bmatrix},
\qquad
v =
\begin{bmatrix}
2\\
-1
\end{bmatrix}
$$

The dot product is calculated as:

$$
u \cdot v = (1)(2) + (2)(-1) = 2 - 2 = 0
$$

We can use Python to calculate the dot product and verify that these vectors are orthogonal.

```python
# Define two vectors
u = np.array([1, 2])
v = np.array([2, -1])

# Calculate the dot product
dot_product = np.dot(u, v)

print("u =", u)
print("v =", v)
print("Dot product =", dot_product)
```

### Interpretation

The dot product of $u$ and $v$ is 0. Therefore, the two vectors are orthogonal.

## 1.2.3 Gram-Schmidt Process

The Gram-Schmidt process converts a set of linearly independent vectors into a set of orthogonal vectors.

Consider the following vectors:

$$
v_1 =
\begin{bmatrix}
1\\
1
\end{bmatrix},
\qquad
v_2 =
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

First, let:

$$
u_1 = v_1
$$

Then subtract the projection of $v_2$ onto $u_1$:

$$
u_2 = v_2 - \frac{v_2 \cdot u_1}{u_1 \cdot u_1}u_1
$$

We can use Python to perform the Gram-Schmidt process and verify that the resulting vectors are orthogonal.

```python
# Define the original vectors
v1 = np.array([1.0, 1.0])
v2 = np.array([1.0, 0.0])

# First orthogonal vector
u1 = v1

# Calculate the projection of v2 onto u1
projection = (np.dot(v2, u1) / np.dot(u1, u1)) * u1

# Subtract the projection from v2
u2 = v2 - projection

# Verify that u1 and u2 are orthogonal
dot_product = np.dot(u1, u2)

print("u1 =", u1)
print("u2 =", u2)
print("u1 dot u2 =", dot_product)
```

### Interpretation

The Gram-Schmidt process produced the vectors $u_1 = (1,1)$ and $u_2 = (0.5,-0.5)$. Their dot product is 0, which verifies that the resulting vectors are orthogonal.

## 1.2.4 Eigenvalues and Eigenvectors

An eigenvector of a matrix is a nonzero vector whose direction does not change when the matrix is applied to it. The corresponding eigenvalue tells how much the eigenvector is scaled.

Consider the matrix:

$$
A =
\begin{bmatrix}
2 & 1\\
1 & 2
\end{bmatrix}
$$

Eigenvalues and eigenvectors satisfy the equation:

$$
Av = \lambda v
$$

We can use Python to calculate the eigenvalues and eigenvectors of $A$, and then verify the relationship $Av = \lambda v$.

```python
# Define the matrix
A = np.array([
    [2, 1],
    [1, 2]
])

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:")
print(A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

```python
# Select the first eigenvalue and eigenvector
lambda1 = eigenvalues[0]
v = eigenvectors[:, 0]

# Calculate both sides of Av = lambda*v
Av = A @ v
lambda_v = lambda1 * v

print("Av =", Av)
print("lambda*v =", lambda_v)

# Verify that the two results are equal
print("Verification:", np.allclose(Av, lambda_v))
```

### Interpretation

The matrix has eigenvalues 3 and 1 with corresponding eigenvectors. The verification returns `True`, showing that the selected eigenvector satisfies $Av = \lambda v$. This confirms the eigenvalue-eigenvector relationship.
