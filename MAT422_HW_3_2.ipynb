## 3.2.1 Limits and Continuity

A limit describes the value that a function approaches as the input approaches a particular point.

The limit of a function is written as

$$
\lim_{x \to a} f(x) = L.
$$

A function is continuous at $x=a$ if

$$
\lim_{x \to a} f(x) = f(a).
$$

Consider the function

$$
f(x) = \frac{x^2-4}{x-2}.
$$

At $x=2$, the function is undefined because the denominator equals zero.

However, factoring the numerator gives

$$
f(x) = \frac{(x-2)(x+2)}{x-2}=x+2,
\quad x\neq 2.
$$

Therefore,

$$
\lim_{x \to 2} f(x)=4.
$$

We can use Python to calculate this limit and graph the function to demonstrate the removable discontinuity at $x=2$.

```python
# Define the symbolic variable
x = sp.symbols('x')

# Define the function
f = (x**2 - 4) / (x - 2)

# Calculate the limit as x approaches 2
limit_value = sp.limit(f, x, 2)

# Simplify the function
simplified_f = sp.simplify(f)

print("Original function:")
print(f)

print("\nSimplified function:")
print(simplified_f)

print("\nLimit as x approaches 2:")
print(limit_value)
```

```python
# Create x-values around x = 2
x_values = np.linspace(0, 4, 400)

# Calculate function values using the simplified expression
y_values = x_values + 2

# Plot the function
plt.plot(x_values, y_values, label="f(x) = (x² - 4)/(x - 2)")

# Show the removable discontinuity at (2, 4)
plt.scatter(2, 4, facecolors="white",
            edgecolors="red", s=100,
            zorder=5, label="Hole at (2, 4)")

# Add labels and title
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Limit and Removable Discontinuity")

plt.legend()
plt.grid()

plt.show()
```

### Interpretation

The limit of the function as $x$ approaches 2 is **4**. Although the original function is undefined at $x=2$, simplifying the expression gives $f(x)=x+2$ for $x\neq 2$. The graph shows a removable discontinuity at $(2,4)$, represented by an open circle. Therefore, the limit exists, but the function is not continuous at $x=2$ because $f(2)$ is undefined.

## 3.2.2 Derivatives

A derivative measures the instantaneous rate of change of a function. Geometrically, the derivative represents the slope of the tangent line at a particular point.

The derivative is defined as

$$
f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x)}{h}.
$$

Consider the polynomial function

$$
f(x)=x^2+2x+1.
$$

Using the power rule, its derivative is

$$
f'(x)=2x+2.
$$

At $x=2$, the derivative is

$$
f'(2)=2(2)+2=6.
$$

The function value at this point is

$$
f(2)=2^2+2(2)+1=9.
$$

Therefore, the tangent line at $(2,9)$ has slope 6.

Using the point-slope equation,

$$
y-9=6(x-2),
$$

we obtain

$$
y=6x-3.
$$

We can use Python to calculate the derivative and verify the tangent line at $x=2$.

```python
# Define the symbolic variable
x = sp.symbols('x')

# Define the function
f = x**2 + 2*x + 1

# Calculate the derivative
f_prime = sp.diff(f, x)

# Evaluate the function and derivative at x = 2
x_point = 2
y_point = f.subs(x, x_point)
slope = f_prime.subs(x, x_point)

# Calculate the tangent line
tangent_line = slope * (x - x_point) + y_point

print("Original function:")
print(f)

print("\nDerivative:")
print(f_prime)

print("\nFunction value at x = 2:")
print(y_point)

print("\nSlope at x = 2:")
print(slope)

print("\nTangent line:")
print(sp.expand(tangent_line))
```

```python
# Create x-values for plotting
x_values = np.linspace(-2, 5, 400)

# Calculate the function values
y_values = x_values**2 + 2*x_values + 1

# Calculate the tangent line values
tangent_values = 6*x_values - 3

# Plot the original function
plt.plot(x_values, y_values, label="f(x) = x² + 2x + 1")

# Plot the tangent line
plt.plot(x_values, tangent_values,
         linestyle="--", label="Tangent line: y = 6x - 3")

# Mark the point of tangency
plt.scatter(2, 9, color="red", s=80,
            zorder=5, label="Point (2, 9)")

# Add labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title("Derivative and Tangent Line")

plt.legend()
plt.grid()

plt.show()
```

### Interpretation

The derivative of the function $f(x)=x^2+2x+1$ is $f'(x)=2x+2$. At $x=2$, the derivative equals **6**, meaning the instantaneous rate of change is 6 at that point. The tangent line is $y=6x-3$, which passes through $(2,9)$. The graph confirms that the tangent line has the same slope as the function at this point, demonstrating how derivatives describe instantaneous rates of change.

## 3.2.3 Taylor's Theorem

Taylor's theorem allows us to approximate a differentiable function using a polynomial constructed from its derivatives at a particular point.

The Taylor polynomial of degree $n$ centered at $a$ is

$$
T_n(x)=\sum_{k=0}^{n}\frac{f^{(k)}(a)}{k!}(x-a)^k.
$$

For a function with sufficiently many derivatives, Taylor's theorem also provides a remainder term describing the approximation error.

Consider the function

$$
f(x)=e^x.
$$

Since every derivative of $e^x$ is also $e^x$, all derivatives evaluated at $x=0$ equal 1.

The Taylor polynomial centered at $a=0$ is called a Maclaurin polynomial.

The first few approximations are

$$
T_1(x)=1+x,
$$

$$
T_2(x)=1+x+\frac{x^2}{2},
$$

and

$$
T_3(x)=1+x+\frac{x^2}{2}+\frac{x^3}{6}.
$$

We can use Python to calculate these Taylor polynomials and compare their approximations with the original function.

```python
# Define the symbolic variable
x = sp.symbols('x')

# Define the function
f = sp.exp(x)

# Calculate Taylor polynomials centered at x = 0
T1 = sp.series(f, x, 0, 2).removeO()
T2 = sp.series(f, x, 0, 3).removeO()
T3 = sp.series(f, x, 0, 4).removeO()

print("Original function:")
print(f)

print("\nFirst-degree Taylor polynomial:")
print(T1)

print("\nSecond-degree Taylor polynomial:")
print(T2)

print("\nThird-degree Taylor polynomial:")
print(T3)

# Compare approximations at x = 1
actual_value = float(f.subs(x, 1))

print("\nActual value of e^1:")
print(round(actual_value, 6))

print("\nTaylor approximations at x = 1:")
print("T1 =", float(T1.subs(x, 1)))
print("T2 =", float(T2.subs(x, 1)))
print("T3 =", float(T3.subs(x, 1)))
```

```python
# Create x-values for plotting
x_values = np.linspace(-2, 2, 400)

# Calculate the original function
y_actual = np.exp(x_values)

# Convert Taylor polynomials to numerical functions
T1_func = sp.lambdify(x, T1, "numpy")
T2_func = sp.lambdify(x, T2, "numpy")
T3_func = sp.lambdify(x, T3, "numpy")

# Calculate Taylor polynomial values
y_T1 = T1_func(x_values)
y_T2 = T2_func(x_values)
y_T3 = T3_func(x_values)

# Plot the original function
plt.plot(x_values, y_actual, label="f(x) = e^x")

# Plot the Taylor polynomial approximations
plt.plot(x_values, y_T1, label="T1 (Degree 1)")
plt.plot(x_values, y_T2, label="T2 (Degree 2)")
plt.plot(x_values, y_T3, label="T3 (Degree 3)")

# Add labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title("Taylor Polynomial Approximations of e^x")

plt.legend()
plt.grid()

plt.show()
```

### Interpretation

Taylor's theorem allows us to approximate a function using polynomials based on its derivatives. For the function $f(x)=e^x$, the first three Taylor polynomials centered at $x=0$ were calculated and compared with the original function.

At $x=1$, the actual value is approximately **2.7183**, while the first-, second-, and third-degree approximations are **2.0**, **2.5**, and **2.6667**, respectively.

The graph shows that higher-degree Taylor polynomials provide increasingly accurate approximations near the center $x=0$. The third-degree polynomial follows the exponential function more closely than the lower-degree polynomials in the region shown near the center. This demonstrates how Taylor polynomials can approximate functions with increasing accuracy as additional terms are included.
